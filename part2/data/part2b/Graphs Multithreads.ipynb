```python
import matplotlib.pyplot as plt

# data
threads = [1, 2, 4, 8]              # x-axis
runtimes = [108.135, 55.232, 28.702, 22.535]  # y-axis (seconds)

# Create the plot
plt.figure()
plt.plot(threads, runtimes, marker='o', label='barnes runtime')

# Labels and title
plt.xlabel('Number of Threads')
plt.ylabel('Runtime (seconds)')
plt.title('Benchmark: barnes')

# Show legend
plt.legend()

# Optional: grid for readability
plt.grid()

# Show the plot
plt.show()

print("Speedup")
T1 = runtimes[0]
speedup = [T1 / t for t in runtimes]
print(speedup)


print("Efficiency")
threads = [1, 2, 4, 8]
efficiency = [s / n for s, n in zip(speedup, threads)]
print(efficiency)



ideal = [T1 / n for n in threads]

plt.plot(threads, runtimes, marker='o', label='Actual')
plt.plot(threads, ideal, linestyle='--', label='Ideal')

plt.legend()
plt.show()

```
